Car la célérité $c$ n'est pas une vitesse, c'est une constante, c'est-à-dire un artifice math&physique qui permet d'entériner tout débat à son propos. En maths une vitesse est une dérivé : Dans le cas de $c$, ça devrait être $2c$, l'accélération est : $c^2$, la dérivé : $2c$ et la position : $c\pm=\frac{2c}{2}$
