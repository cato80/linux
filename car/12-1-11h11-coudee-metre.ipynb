```python
def f(x):
    if x<0:
        return sqrt(-x^2-2*x)
    else:
        return -sqrt(-x^2+2*x)
a=plot(f,(x,-2,2),color='white',
       fill=sqrt(4-x^2),fillcolor="lightgray",alpha=0.1)
b=plot(f,(x,-2,2),color='white',
       fill=-sqrt(4-x^2),fillcolor="white",alpha=0.1)
p = circle((0,0),2,color="yellow",zorder=-5)
show(a+p,figsize=2,axes=False,aspect_ratio=1)
```
