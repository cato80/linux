###### Nouvelle approche de la physique

La terminologie qu'utilise la physique est de plus en plus obsolète, on ne devrait plus parler de masse en nanophysique, mais de rapport.

Pour cela il faut définir plus précisèment la notion de particule : Je définie une particule comme un rapport, je fais l'analogie avec la musique. On a ainsi un intervalle et une note la particule. Quand j'accorde ma guitare, je me suis rendu compte que je devais coupler 2 sons, un diéze et un bémol, qui lorsqu'elle se trouve en résonance (équilibre) produise le son désiré, le son est une compression de 2 intervalles. C'est ce que la science appelle l'onde. Les 2 intervalles sont des milieux différents, ils produisent la note à la résonance (couplage).
