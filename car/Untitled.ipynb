```python
%matplotlib inline
import numpy as np
from pylab import *
from qutip import *
```

```python
b = Bloch();b.show()
```

```python
pnt = [1/np.sqrt(3),1/np.sqrt(3),1/np.sqrt(3)]
b.add_points(pnt);b.show()
```

```python
vec = [0,1,0];b.add_vectors(vec);b.show()
```

```python
up = basis(2,0);b.add_states(up);b.show()
```

```python
b = Bloch();pnt = [1/np.sqrt(3),1/np.sqrt(3),1/np.sqrt(3)]
b.add_points(pnt);vec = [0,1,0];b.add_vectors(vec)
up = basis(2,0);b.add_states(up);b.show()
```

```python
x = (basis(2,0)+(1+0j)*basis(2,1)).unit()
y = (basis(2,0)+(0+1j)*basis(2,1)).unit()
z = (basis(2,0)+(0+0j)*basis(2,1)).unit()
b.add_states([x,y,z]);b.show()
```
