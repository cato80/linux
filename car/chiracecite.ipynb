```python
def complexpoints_plot(pts):
    return list_plot([(real(i), imag(i)) for i in pts],
    axes_labels = ["Re($z$)", "Im($z$)"],size=25)
complexpoints_plot([2*I, e^(I*pi),e^(I*2*pi/3), 2-1*I])
pts = [e^(i*(theta))
       for theta in srange(0, 2*pi, .1)]
complexpoints_plot(pts)
```
