Les scientifiques font une grosse erreur : Les particules n'ont un nom propre que dans le cas de remembrement, c'est un changement d'échelle , un électron est un photon à une échelle différente, dans un état différent. A l'origine, au début la particule est double, couplée, comme un oeuf : blanc/jaune. C'est ce qu'exprime le dao, la voie. Qui en passant n'est pas d'origine chinoise, mais bien plus antique.

A partir de la vitesse de la lumière (qui n'en est pas une), on ne devrait plus parler de masse mais de résistance, d'impédance, qui est une force de freinage. Dans $E=mc^2$, la masse est une impédance qui freine la vitesse du photon, mais comme la célérité est supraluminique $c^2$, les particules explosent, produisant une espèce de magma, l'énergie qu'il vaudrait mieux appelé plasma.

Le mot éther est le plus approprié pour parler du vide : Le tort des scientifiques c'est ne pas avoir compris ce qu'il était !

L'éther c'est l'impédance du vide : E$\Omega$. Le vide n'est pas vide # 0, c'est le milieu : éther. Son centre est vide, donc signature différente du milieu : [-1 0 1 ], qu'on peut remplacer par -0+ (trinaire balancé : tit ).
