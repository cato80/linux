#####  Longueur d'onde fondamentale $\lambda$ = $2 \pi$
$\lambda=\frac{\frac{1}{\sqrt\mu\epsilon}}{\frac{1}{2\pi\sqrt\mu\epsilon}}$ = $\frac{c}{\nu}$=$2\pi$
