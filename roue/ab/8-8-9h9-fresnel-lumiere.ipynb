Fresnel,  en  1818,  fait  l'hypothèse selon  laquelle  on  peut  considérer  le retard de la lumière dans un milieu plus réfringent que le vide lorsque ce milieu est  immobile,  comme  résultant  uniquement  d'une  plus  grande  densité  

Zéro est immobile, donc plus réfringent que le vide, le vide n'est pas vide, il a une impédance de 360 $\Omega$ (valeur que j'ai calculé) ou $2 \pi$, un cercle complet. La célérité $c$ est freiné à 376 $\Omega$ environ, l'impédance du vide. C'est pourquoi il ne faut pas appeler $c$ vitesse de la lumière,( tout au plus vitesse d'une particule ou du photon), la lumière est la fréquence de résonance du dipole L(inductance=perméabilité, état de retenir,réaction) et  C (capacitance=permittivité, état de libérer, action), le yin (L) yang (C) en est la plus parfaite représentation. La voie est la lumière. La relativité est une théorie pseudo géométrique. Donc on doit pouvoir valider la notion de force avec le déterminisme géométrique.
