##### Calcul du rayon classique de l'électron

$r_e=\frac{1}{4\pi \epsilon_0}$ $\frac{e^2}{m c^2}$ = $2.817940325 \times 10^{-15} m$ 

Le premier rapport est $k$ coefficient de Coulomb.

```python
mu=4*3.141592653E-7;c=299792458;ep=1/(mu*c*c)
e=1.60E-19;m=9.10E-31;k=1/(ep*(mu/1E-7))
Re=(1/k)*(e^2/m*c^2);Re=2.817940E-15
```

###### Constante de Hubble

```python
a=1/137.0;B=3-sqrt(5.0);pars=96939420213600000000/n(pi)
H0=(c*(a^19)*(sqrt(B)))/Re;Hu=H0*(pars)
show(LatexExpr(r"H_0=\frac{c a^{19} \sqrt B}{r_e}="),H0,"m")
show(LatexExpr(r"H_0\frac{MParsec}{1000}="),Hu," km/s");
```
