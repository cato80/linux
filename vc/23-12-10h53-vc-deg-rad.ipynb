L'impédance du vide (éther) induit une perte de vitesse, étant l'inverse de la conductance(c=v):

$Z=z^2=\frac {\mu}{\epsilon}$

Qu'il vaut mieux écrire pour la simplicité LC :

L=inductance (tension) C=capacitance (intensité) et $Z=z^2=\frac {L}{C}$, Z étant l'impédance du vide (qui n'est pas vide = Z )
