```python
Cc=1/(4*n(pi)*ep);Cc#  7 décimales -6 pour kg
```

```python
int(cc/Cc)
```

On voit que Coulomb a mesuré une particule $kc$, inférieure de un dix millionième du photon. C'était déjà une prouesse que de la mesurer, aujourd'hui ce ne serait pas évident de le faire avec tout la pollution, le bruit et les tremblements !
